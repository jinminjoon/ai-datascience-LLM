## partial_variables(부분변수 채움)

```python
from datetime import datetime

# 오늘 날짜
datetime.now().strftime("%B %d")
```

```python
# 날짜를 변환하는 함수 정의
def get_today():
  return datetime.now().strftime('%B %d')
```

```python
prompt = PromptTemplate(
    template= "오늘의 날짜는 {today} 입니다. 오늘이 기념일인 유명인 {n}명을 알려주세요. 기념일을 표시해주세요.",
    input_variables=["n"],
    partial_variables={
        "today":get_today
    }
)
prompt
```

```python
prompt.format(n=3)
```

```python
# chain 생성
chain = prompt | llm
```

```python
print(chain.invoke(3).content)
```

```python
print(chain.invoke({"today":"oct 30", "n":3}).content)
```

## ChatPromptTemplate
  - 대화형식의 프롬프트로 구성
  - "system": 시스템 설정 메시지
  - "human": 사용자
  - "ai" : AI의 답변 메세지

```python
from langchain_core.prompts import ChatPromptTemplate

# 수정된 코드: 바깥쪽 중괄호{}를 제거했습니다.
chat_prompt = ChatPromptTemplate.from_template("{player}의 소속팀은 어디인가요?")

chat_prompt
```

```python
chat_prompt.format(player="비니시우스")
```

```python
chat_template = ChatPromptTemplate.from_messages(
    [
        #role, message
        ("system", "당신은 친절한 AI 어시스턴트입니다. 당신의 이름은 {name} 입니다."),
        ("human", "반가워요!"),
        ("ai", "안녕하세요! 무엇을 도와드릴까요?"),
        ("human", "{user_input}"),
    ]
)

# 챗 message를 생성합니다.
messages = chat_template.format_messages(
    name="mj", user_input="당신의 성함은 무엇입니까?"
)
messages
```

## MessagePlaceholder
  - 메세지 프롬프트 템플릿에 어떤 역할을 사용해야 할지 확실치 않거나 서식 지정중 메세지 목록을 삽입할때 유용하다

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

chat_prompt = ChatPromptTemplate.from_messages(
    [
        MessagesPlaceholder(variable_name="conversation"),
        ("human", "지금까지의 대화를 {sentence_count}개 문장으로 요약합니다."),
    ]
)
chat_prompt
```

```python
chain = chat_prompt | llm | StrOutputParser()
```

```python
chat_prompt.format(
    sentence_count=3,
    conversation=[
        ("human", "안녕하세요! 이번시즌 새로 이적한 비니시우스 입니다. 저는 이번시즌 득점왕이 목표이고 팀 우승에도 기여하고 싶네요."),
        ("ai", "오, 멋진 포부를 가지고 있으시네요. 대략 몇골정도 예상하시나요?."),
        ("human", "30골이상 목표를 하고 있습니다."),
        ("ai", "멋진 포부이시네요. 화이팅!")
    ]
)
```

```python
# chain 실행 및 결과확인
chain.invoke(
    {
        "player": "비니시우스",
        "sentence_count":3,
        "conversation":[
        ("human", "안녕하세요! 이번시즌 새로 이적한 비니시우스 입니다. 저는 이번시즌 득점왕이 목표이고 팀 우승에도 기여하고 싶네요."),
        ("ai", "오, 멋진 포부를 가지고 있으시네요. 대략 몇골정도 예상하시나요?."),
        ("human", "30골이상 목표를 하고 있습니다."),
        ("ai", "멋진 포부이시네요. 화이팅!")
    ],
  }
)
```

## 퓨샷프롬프트(FewshotPromptTemplate)

```python
from langchain_core.prompts.few_shot import FewShotPromptTemplate
from langchain_core.prompts import PromptTemplate

examples = [
    {
        "question": "마라도나와 펠레 중 누가 더 오래 살았나요?",
        "answer": """이 질문에 추가로 질문이 더 필요한가요?: 예.
추가 질문: 마라도나는 몇 살에 사망했나요?
중간 답변: 마라도나는 60세에 사망했습니다.
추가 질문: 펠레는 몇 살에 사망했나요?
중간 답변: 펠레는 82세에 사망했습니다.
최종 답변은: 펠레
"""
    },
    {
        "question": "삼성 창립자는 언제 태어났나요?",
        "answer": """이 질문에 추가 질문이 필요한가요?: 예.
추가 질문: 삼성의 창립자는 누구인가요?
중간 답변: 삼성은 이병철에 의해 창립되었습니다.
추가 질문: 이병철은 언제 태어났나요?
중간 답변: 이병철은 1910년 2월 12일에 태어났습니다.
최종 답변은: 1910년 2월 12일
""",
    },
    {
        "question": "율곡 이이의 어머니가 태어난 해의 통치하던 왕은 누구인가요?",
        "answer": """이 질문에 추가 질문이 필요한가요: 예.
추가 질문: 율곡 이이의 어머니는 누구인가요?
중간 답변: 율곡 이이의 어머니는 신사임당입니다.
추가 질문: 신사임당은 언제 태어났나요?
중간 답변: 신사임당은 1504년에 태어났습니다.
추가 질문: 1504년에 조선을 통치한 왕은 누구인가요?
중간 답변: 1504년에 조선을 통치한 왕은 연산군입니다.
최종 답변은: 연산군
"""
    }
]
```

```python
example_prompt = PromptTemplate.from_template(
    "Question:\n{question}\nAnswer:\n{answer}"
)

print(example_prompt.format(**examples[0]))
```

```python
prompt = FewShotPromptTemplate(
    examples=examples,
    example_prompt=example_prompt,
    suffix="Question:\n{question}\nAnswer:",
    input_variables=["question"],
)

question = "이강인이 psg에 이적한 연도에 이강인의 나이는 몇 살인가요?"
final_prompt = prompt.format(question=question)
print(final_prompt)
```

## Example Selector
  - 프롬프트 예제가 많은 경우 프롬프트에 포함할 예제를 선택
  - (질문과 유사도 검색을해서 유사도가 높은 예제를 선택)

```python
!pip install -U langchain-chroma
```

```python
from langchain_core.example_selectors import SemanticSimilarityExampleSelector
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings

# Vector DB 생성(저장소 이름, 임베딩 클래스)
chroma = Chroma("example_selector", OpenAIEmbeddings())

example_selector = SemanticSimilarityExampleSelector.from_examples(
    # 예시들
    examples,
    # 의미적 유사성을 확인하기 위한 임베딩
    OpenAIEmbeddings(),
    # 임베딩 저장 및 유사성 검색
    Chroma,
    # 생성할 예시의 수
    k=1,
)

# 입력과 가장 유사한 예시를 선택합니다.
selected_examples = example_selector.select_examples({"question":question})

question = "김민재가 바이에른 뮌헨에 이적한 연도에 김민재의 나이는 몇 살인가요?"
print(f"입력에 가장 유사한 예시:\n{question}\n")
for example in selected_examples:
  print(f"question:\n{example["question"]}")
  print(f"answer:\n{example["answer"]}")
```

```python
from langchain_core.output_parsers import StrOutputParser

prompt = FewShotPromptTemplate(
    example_selector=example_selector,
    example_prompt = example_prompt,
    suffix = "Question:\n{question}\nAnswer:",
    input_variables=["question"],
)

question = "김민재가 바이에른 뮌헨에 이적한 연도에 김민재의 나이는 몇 살인가요?"
example_selector_prompt = prompt.format(question=question)

# 체인
chain = prompt | llm | StrOutputParser()
chain.invoke({"question":"김민재가 바이에른 뮌헨에 이적한 연도에 김민재의 나이는 몇 살인가요?"})
```
