## 템플릿 변경

```python
template = """
당신은 초등학생을 가르치는 영어 선생님입니다. 주어진 상황에 맞는 영어 회화를 작성하세요.
양식은 [FORMAT]을 참고해서 작성해 주세요.

상황:
{question}

FORMAT:
-영어 회화:
-한글 해석:
"""

# 프롬프트 템플릿을 이용하여 프롬프트 생성
prompt = PromptTemplate.from_template(template)

model = ChatOpenAI(model_name='gpt-4-turbo')

# 문자열 출력 파서
output_parser = StrOutputParser()
```

```python
# 체인 구성
chain = prompt | model | output_parser
```

```python
print(chain.invoke({"question":"저는 백화점에서 음식을 주문하고 싶어요."}))
```

```python
print(chain.invoke({"question": "독일에서 소시지 주문"}))
```

## batch
  - 여러개 딕셔너리를 포함한 리스트를 인자로 받아 일괄처리를 수행


```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

model = ChatOpenAI()
prompt = PromptTemplate.from_template("{topic} 에 대하여 5문장으로 설명해줘.")
chain = prompt | model | StrOutputParser()
```

```python
chain.batch([{"topic":"instagram"},{"topic":"facebook"}])
```

```python
chain.batch(
    [
        {"topic":"머신러닝"},
        {"topic":"딥러닝"},
        {"topic":"casual inference"},
    ],
    config={"max_concurrency":4}
)
```

## 비동기 메소드
  - 프로그램이 특정한 작업을 수행하고 있을 때, 그작업이 완료될 때까지 프로그램을 멈추지 않고 다른 작업을 같이 계속 수행할 수 있게 해주는 방법

```python
# 비동기 스트림 사용
async for token in chain.astream({"topic":"instagram"}):
  print(token, end="", flush=True)
```

```python
mj_process = chain.ainvoke({"topic":"apple"})
```

```python
await mj_process
```

```python
mj_abatch_process = chain.abatch(
    [{"topic":"박지성"}, {"topic":"차범근"},{"topic":"ronaldo"}]
)
```

```python
await mj_abatch_process
```

## Parallel(병렬)

```python
from langchain_core.runnables import RunnableParallel

#화승기업 위치 질문
chain1 = (
    PromptTemplate.from_template("{company} 의 위치는 어디야?")
    | model
    | StrOutputParser()
)

#화승기업 아이템 질문
chain2 = (
    PromptTemplate.from_template("{company} 의 주요 아이템이 무엇이야")
    | model
    | StrOutputParser()
)

combined = RunnableParallel(area=chain1, item=chain2)
```

```python
combined.invoke({"company":"화승코퍼레이션"})
```

```python
prompt = PromptTemplate.from_template("{num} 의 5배는?")
llm = ChatOpenAI(temperature=0)

chain = prompt | llm
```

## RunnablePassthrough.assign()
   - 입력 값으로 들어온 값의 key-value 쌍과 새롭게 할당된 key-value 쌍을 합침

```python
(RunnablePassthrough.assign(new_num=lambda x: x["num"]*5)).invoke({"num":2})
```

```python
runable = RunnableParallel(
    passed = RunnablePassthrough(),
    extra=RunnablePassthrough.assign(mult=lambda x: x["num"] *5),
    modified=lambda x: x["num"] +1,
)

runable.invoke({"num": 2})
```
